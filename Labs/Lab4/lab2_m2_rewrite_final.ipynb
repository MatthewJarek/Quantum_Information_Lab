<h5>3.3 Python Simulation</h5>

```python
import numpy as np

shots = 1024

## Simulate Pauli-X Flip gate ##

Hgate_outcomes = np.random.choice(["|0>","|1>"], shots, p = [0,1])
kets, counts = np.unique(Hgate_outcomes, return_counts = True)

# Xgate_outcomes = np.random.choice(["|0>","|1>"], shots, p = [0,1])
# kets, counts = np.unique(Xgate_outcomes, return_counts = True)
print("==== Circuit 2 X-Gate ====")
print(f"|0> : {0 : .2%}, |1> : {1 : .2%}")
```
