```python
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram, plot_bloch_multivector
import matplotlib.pyplot as plt
import numpy as np
simulator = AerSimulator()
print('Qiskit ready!')
```

###Part 1
------

```python
qc_super = QuantumCircuit(1,1)
qc_super.h(0)
qc_super.measure(0,0)
qc_super.draw(output = 'latex')
```

```python
shots_super = 1024
job_super = simulator.run([qc_super], shots = shots_super)
plot_histogram(job_super.result().get_counts(0))
```

expect |a|^2 = .5, |b|^2 = .5

```python
super_counts = job_super.result().get_counts(0)

print(f" |a|^2 = {super_counts['0']/shots_super : .2%}, |b|^2 = {super_counts['1']/shots_super : .2%}")
```

Close enough

###Part 2
------

```python
qc_bell = QuantumCircuit(2,2)
qc_bell.h(0)
qc_bell.cx(0,1)
qc_bell.measure(0,0)
qc_bell.measure(1,1)
qc_bell.draw(output = 'latex')
```

```python
shots_bell = 1024
job_bell = simulator.run([qc_bell], shots = shots_bell)
plot_histogram(job_bell.result().get_counts(0))
```

(counts(00) + counts (11))/ total = 100% (perfect world)

```python
bell_counts = job_bell.result().get_counts(0)

print(f" |a|^2 = {bell_counts['00']/shots_bell : .2%}, |b|^2 = {bell_counts['11']/shots_bell : .2%} -> |a|^2 + |b|^2 = {bell_counts['00']/shots_bell + bell_counts['11']/shots_bell : .2%} ")
```

###Part 3
------

```python
qc_mod = QuantumCircuit(2,2)
qc_mod.x(1)
qc_mod.h(0)
qc_mod.cx(0,1)
qc_mod.measure(0,0)
qc_mod.measure(1,1)
qc_mod.draw(output="latex")
```

Based on the circuit, q1 will now be in |1> state instead of |0> state when at the cnot gate. This will mean the two states would be |01> and |10>.

```python
shots_mod = 1024
job_mod = simulator.run([qc_mod], shots = shots_mod)
plot_histogram(job_mod.result().get_counts(0))
```

Yes

```python
qc_design = QuantumCircuit(1,1)
qc_design.rx(np.deg2rad(60),0)
plot_bloch_multivector(qc_design)
```

```python
from qiskit.quantum_info import Statevector
```

```python
state = Statevector.from_instruction(qc_design)
print(f"{np.real(state[0]**2): .2%} -> |0> + {np.real(state[1]**2): .2%} -> |1>") 
```

```python
print(np.cos(np.deg2rad(60))**2)
print(np.sin(np.deg2rad(60))**2)
```
