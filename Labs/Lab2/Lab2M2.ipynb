```python
import numpy as np

shots = 1024

## Simulate Circuit 1 H-gate ##

Hgate_outcomes = np.random.choice(["|0>","|1>"], shots, p = [.5,.5])
kets, counts = np.unique(Hgate_outcomes, return_counts = True)
print("==== Circuit 1 H-Gate ====")
print(f"{kets[0]} : {counts[0]/shots : .2%}, {kets[1]} : {counts[1]/shots : .2%}")

# Xgate_outcomes = np.random.choice(["|0>","|1>"], shots, p = [0,1])
# kets, counts = np.unique(Xgate_outcomes, return_counts = True)
print("==== Circuit 2 X-Gate ====")
print(f"{kets[0]} : {0/shots : .2%}, {kets[1]} : {shots/shots : .2%}")


Bell_outcomes = np.random.choice(["|00>","|11>"], shots, p = [.5,.5])
kets, counts = np.unique(Bell_outcomes, return_counts = True)
print("==== Circuit 3 Bell ====")
print(f"{kets[0]} : {counts[0]/shots : .2%}, {kets[1]} : {counts[1]/shots : .2%}")
```

<h5><b>Key Question:</b> Why only |00&rangle; + |11&rangle;?</h5>
<p>The two states are fully entagled, leading to only two results, if the first bit is measured as 1 the second one has to be 1 as well. Theoretically this works, but in the actual run there are errors that prevent this from working consistantly.</p>

<h4>Discussion Questions</h4>
<p><b>Why do histograms vary between runs?</b> The runs are for 1024 shots. Each run is independent on other runs, to ensure a 50/50 split we would have many more shots.</p>
<p><b>Circuit 3: Why no |01&rangle; or |10&rangle; states?</b> The expected output for Circuit 3 states that the only 2 outputs are |00&rangle; or |11&rangle;. This is due to entanglement as talked about above.</p>
<p><b>How many shots improve statistics?</b>This depends on the starting number of shots. Here we have 1024, which gets us pretty close to average. The closer you get to the expected outcome the more shots you have to add (exponential growth). Below I messed with the number of shots and observed that running it with 4096 shots lead to noticibly more stable results.</p>
