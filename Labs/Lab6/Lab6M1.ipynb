```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, state_fidelity
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_bloch_multivector, plot_state_qsphere, plot_histogram
import matplotlib.pyplot as plt
import numpy as np

sim = AerSimulator()
print("Qiskit visualization tools ready!")
```

```python
# |0> state
sv0 = Statevector.from_label('0')
print(f"|0>: {sv0.data}")
plot_bloch_multivector(sv0)
```

```python
# |+> = H|0>
qc_h = QuantumCircuit(1)
qc_h.h(0)
sv_plus = sv0.evolve(qc_h)
print(f"|+>: {sv_plus.data}")
```

```python
plot_bloch_multivector(sv_plus)
```

```python
# X gate: North -> South pole

qc_x = QuantumCircuit(1)
qc_x.x(0)
sv_x = sv0.evolve(qc_x)
```

```python
plot_bloch_multivector(sv0, title="X Gate Before")
```

```python
plot_bloch_multivector(qc_x, title="X Gate After")
```

```python
# H|0> vs H|1> phase difference
qc_h0 = QuantumCircuit(1)
qc_h0.h(0)
sv_h0 = Statevector.from_instruction(qc_h0)
qc_h1 = QuantumCircuit(1)
qc_h1.x(0)
qc_h1.h(0)
sv_h1 = Statevector.from_instruction(qc_h1)
```

```python
plot_state_qsphere(sv_h0.data)
```

```python
plot_state_qsphere(sv_h1.data)
```

```python
# Bell pair statevector
qc_bell = QuantumCircuit(2)
qc_bell.h(0)
qc_bell.cx(0,1)
sv_bell = Statevector.from_label('00').evolve(qc_bell)
print(f"Bell: {sv_bell}")
plot_state_qsphere(sv_bell.data)
```
