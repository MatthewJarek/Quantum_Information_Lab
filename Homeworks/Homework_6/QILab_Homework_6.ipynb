```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, state_fidelity
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_bloch_multivector, plot_state_qsphere, plot_histogram
import matplotlib.pyplot as plt
import numpy as np

sim = AerSimulator()
```

##### Part 1

```python
qc_minus = QuantumCircuit(1)
qc_minus.h(0)
qc_minus.z(0)
sv_minus = Statevector.from_label('0').evolve(qc_minus)
```

```python
plot_bloch_multivector(qc_minus, title="ZH|0⟩ -> |-⟩")
```

For this result I first put the qubit into a super position of 1/sqrt{2} (|0> + |1>) using the Hadamard gate. Then using the Z gate, I flipped the |1> vector from +|1> to -|1> creating the given state 1/sqrt{2} (|0> - |1>)

##### Part 2

By implimenting HZH, I would expect the q-sphere to be in the |1> state (straight down) with no phase (blue) this is because the first two gates put the qubit into |-> (the result from part 1), and I know that H|1> = |-> and that HH = 1, so H|1> = |-> => HH|1> = H|-> => |1> = H|->.

```python
qc_hzh = QuantumCircuit(1)
qc_hzh.h(0)
qc_hzh.z(0)
qc_hzh.h(0)
sv_hzh = Statevector.from_label('0').evolve(qc_hzh)
plot_state_qsphere(sv_hzh.data)
```
