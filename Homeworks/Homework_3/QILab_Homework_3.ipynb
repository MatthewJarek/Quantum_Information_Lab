```python
import numpy as np
shots = 1024

print("==== Circuit 4: Swapped H+CNOT (CNOT -> H) ====")
swapped = np.random.choice(["|00>", "|01>"], size = shots, p = [.5,.5])
unique, counts = np.unique(swapped, return_counts = True)
print(f"*Expected*  {unique[0]}: 50%,    {unique[1]}: 50%")
print(f"*Simulated* {unique[0]}:{counts[0]/shots : .2%}, {unique[1]}:{counts[1]/shots : .2%}")

```
