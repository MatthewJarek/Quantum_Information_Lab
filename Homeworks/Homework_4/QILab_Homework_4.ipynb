```python
import numpy as np

print("Simulating shot experiments with precision analysis\n")
output = []
for i in range(1):
    shots = 3000
    prob = [0.6,0.4]
    outcomes = np.random.choice([0,1], size=shots, p=prob)

    # Measure both proportions (they always sum to 1)
    p1_measured = np.sum(outcomes == 1) / shots
    p0_measured = 1 - p1_measured
    print(f"############# Run #{i+1} #############")
    print(f"Expected: {prob[0] :.1%} |0> and {prob[1] :.1%} |1>")
    print(f"Actual: {p0_measured :.1%} |0> and {p1_measured :.1%} |1>")
    ev = [shots*.6, shots*.4]
    print(f"Number of |0>: {np.sum(outcomes ==0)} (Expected: {ev[0] :1.0f})")
    print(f"Number of |1>: {np.sum(outcomes ==1)} (Expected: {ev[1] :1.0f})")
    sigmap = np.sqrt((.6*.4)/shots)
    sigma = np.sqrt(.6*.4*shots)
    print(f"Standard deviation: {sigma :.1f}")
    print(f"Probability Standard deviation: {sigmap :.5f}\n\n")
    output.append(f"|{shots:^12}|{sigmap :^12.5f}| ±{sigmap*1.96:^8.1%}  | {prob[0]-sigmap*1.96:^.1%} - {prob[0]+sigmap*1.96:^.1%} | ±{sigmap*2.576:^8.1%}  | {prob[0]-sigmap*2.576:^.1%} - {prob[0]+sigmap*2.576:^.1%} |")
    
print(f"-------------------------------------------------------------------------------------")
print(f"| Shots (n)  |     σp     | 95% Margin | Typical Range | 99% Margin | Typical Range |")
print(f"-------------------------------------------------------------------------------------")
for i in range(len(output)):
    print(f"{output[i]}")
    print(f"-------------------------------------------------------------------------------------")

```
